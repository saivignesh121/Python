```python
#list methods
# adding elements to list
#append(ele): Adds the element at the end of list
#extend(list): Adds all elements in the list at end 
l=[1,2,3,4,5]
print(l.append(10))
print(l)
```

```python
#list methods
# adding elements to list
#append(ele): Adds the element at the end of list
#extend(list): Adds all elements in the list at end 
#insert(pos,ele): Insert the element at specified position
lst=[1,2,3,4,5]
lst.append(10)
print(lst)
lst.extend([10,20,30,40,50])
print(lst)
lst.insert(3,100)
print(lst)
lst.insert(100,1000)
print(lst)
lst.insert(-1,200)
print(lst)
```

```python
# Removing Elements from list
#remove(val): it removes the first occurance value from list,otherwise valueerror
#pop():it remove and return end value from list,if list is empty it raises error
#pop(position): it remove and return specified position element from list
#clear():it removes all elements from list
lst=[1,2,3,4,5,3]
lst.remove(3)
print(lst)
lst.pop()
print(lst)
print(lst.pop(2),lst)
print(lst)
#print(lst.pop(120),lst) #index Error: pop index out of range
#print([].pop())#index Error
lst.clear()
print(lst)



```

```python
#count(ele):it returns the total occurance of element in list
l=[1,2,3,4,2,2,3]
print(l.count(2))
print(l.count(10))
```

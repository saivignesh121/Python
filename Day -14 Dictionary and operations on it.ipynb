```python
#dictionary values can access by using keys
details = { 'name':"srinu","batch ":49,"course":"python" ,"name":"babu"}
print(details['name'])
```

```python
#dictionary does not support index and slicing
details = { 'name':"srinu","batch ":49,"course":"python" ,"name":"babu",0:1}
print(details[0])
print(details[1])
```

```python
#update Dictionary value
details = { 'name':"srinu","batch ":49,"course":"python" ,"name":"babu",0:1}
print(details["course"])
```

```python
#update Dictionary value
#dictionary update values and items 
details = { 'name':"srinu","batch ":49,"course":"python" ,"name":"babu",0:1}
details['course']='java'
details['age']=50
print(details)
```

```python
#dictionary methods
#get(key,default_value): it returns the values if the key is present in the dictionary
# otherwise it returns default value 'None'
details = { 'name':"srinu","batch ":49,"course":"python" ,"name":"babu",0:1}
print(details.get("location",'india'))
print(details.get("name",'ram'))
print(details.get("marks"))
print(details)
```

```python
#update(items): Dictionary updates with new items
details = { 'name':"srinu","batch":49,"course":"python" ,"name":"babu",0:1}
new_details ={'course':'java','batch':50 ,'location': 'india'}
details.update(new_details)
print(details)
```

```python
#keys() :It returns the list of dictionary keys
details = { 'name':"srinu","batch ":49,"course":"python" ,"name":"babu",0:1}
print(details.keys())
#values : it returns the lsit of dictionary values
print(details.values())
#items() : it returns the dictionary list of tuple of key _value pairs
print(details.items())
```

```python
#pop(key):It returns specified key values otherwise key error
#popitem(): it returns and removes the last item

details = { 'name':"srinu","batch ":49,"course":"python" ,"name":"babu",0:1,"salary":65000}
print(details.pop("course"))
print(details.popitem())
print(details)
```

# string operations
1. string concatenation( use +)
2. string repetition (use *)
   


```python
# concatination
s1='code'
s2='gnan'
print(s1+s2)
```

```python
# concatination
s1='code'
s2='gnan'
print(s1+" "+s2)
print(s1*2)
print((s1 +" ")*5)
```

```python
#case conversion  functions
s="sky is blue ocean is blue"
print(s.upper())
print(s.lower())
print(s.swapcase())
print(s.title())
print(s.capitalize())
```

```python
#string manipulation function
s="sky is blue ocean is blue"
print(s.replace("blue","red"))
print(s)
print(s.split())
print(s.split(sep="Blue"))
print(s.split(sep="blue"))
```

```python
#diference between index and find
s="sky is blue ocean is blue"
print(s.index("is"))
#print(s.index("z"))#error
s="sky is blue ocean is blue"
print(s.find("is"))
print(s.find("z"))
```

```python
#strip function 
s="          sky is blue ocean is blue              "
print(s.strip())
print(s.rstrip())
print(s.lstrip())
```

```python
#"sep".join(iterable)
lst=['sky', 'is', 'blue', 'ocean', 'is', 'blue']
print("".join(lst))
print(" ".join(lst))
print("@".join(lst))
```

```python
#string checking function
s="sky is blue ocean is blue"
print(s.isupper())
print(s.islower())
print(s.isdigit())
print(s.isalpha())
print(s.isalnum())
print(s.istitle())
```

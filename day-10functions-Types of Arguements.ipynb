```python
#Types of arguements
#positional arguements
def sum(a,b,c):
    return a+b+c
x=int(input())
y=int(input())
z=int(input())
print(sum(x,y,z))
```

```python
#positional arguemnets
def sub(a,b):
    return a-b
x=int(input())
y=int(input())
print(sub(x,y))
print(sub(y,x))
```

```python
#keyword arguements arguemnets
def sub(a,b):
    return a-b
x=int(input())
y=int(input())
print(sub(a=x,b=y))
print(sub(b=y,a=x))
```

```python
#keyword arguements arguemnets
#sum of 3 numbers
def sum(a,b,c):
    return a+b+c
x=int(input())
y=int(input())
z=int(input())
print(sum(a=x,b=y,c=z))
print(sum(b=y,a=x,c=z))
```

```python
#keyword arguements arguemnets
# sum of 4 numbers
def sub(a,b,c,d):
    return a+b+c+d
x=int(input())
y=int(input())
z=int(input())
k=int(input())
print(sub(a=x,b=y,c=z,d=k))
print(sub(d=k,b=y,a=x,c=z))
```

```python
#Default arguements
#sum of 4 numbers
def sum(a=0,b=0,c=0,d=0):
    return a+b+c+d
x=int(input())
y=int(input())
z=int(input())
v=int(input())
print(sum(x,y))
print(sum(x,y,z))
print(sum(x,y,z,v))
```

```python
#multiplication of 4 numbers
def mul(a,b,c=1,d=1):
    return a*b*c*d
x=int(input())
y=int(input())
z=int(input())
v=int(input())
print(mul(x,y))
print(mul(x,y,z))
print(mul(x,y,z,v))
```

```python
#variable length positional arguements
#sum of 4 numbers
def add(x,y=0 ,*args):
    print(type(args))
    print(args)
    s=0
    for num in args:
        s+=num
    return s
print(add(10,20,30,40,50))
print(add(10,20,30,40,50,1,2,3,4,5,6))
```

```python
# variable length keyword arguements
def add(x,y=0,*args,**kwargs):
    print(type(args))
    print(args)
    s=0
    for num in args:
        s+=num
    return s
print(add(10,20,30,40,50,1,2,3,4,5,6,a=10,b=20,c=30))
print(add(10))
print(add(10,-20))
print(add(10,20,30,40,50))
```

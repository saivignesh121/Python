```python
# printing even and odd numbers from the list seperately
lst=list(map(int,input().split()))
even=[]
odd=[]
for num in lst:
    if num%2==0:
        even.append(num)
    else:
        odd.append(num)
print(even)
print(odd)
```

```python
# printing even and odd numbers from the list seperately using inde
lst=list(map(int,input().split()))
even=[]
odd=[]
for i in range(1,len(lst)+1):
    if i %2==0:
        even.append(i)
    else:
        odd.append(i)
print(even)
print(odd)
```

```python
#printing the maimum value
lst=list(map(int,(input().split())))
high_val=lst[0]
for i in range(1,len(lst)):
    if high_val<lst[i]:
        high_val=lst[i]
print(high_val)

    
```

```python
#find the target element
arr=[1,3,5,1,6,3,2,3,4]
target=6
count=0
for i in range(len(arr)):
    if target==arr[i]:
        count+=1
print(count)
```

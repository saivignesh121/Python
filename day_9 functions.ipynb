```python
#function definition
def even_or_odd(x):
    if x%2==0:
        return 'even'
    else:
        return 'odd'
#function calling
res = even_or_odd(11)
print(res)
print(even_or_odd(11))
```

```python
# using print insted of return
def even_or_odd(x):
    if x%2==0:
        print ('even')
    else:
        print ('odd')
    print("program done")
#function calling
res = even_or_odd(11)
print(res)
print(even_or_odd(11))
```

```python
#sum of elements in list
def sum_of_ele(lst):
    total=0
    for num in lst:
        total+=num
    return total
lst=[1,2,3,4,5,6,7]
print(sum_of_ele(lst))
```

```python
#sum of elements in list using index
def sum_of_ele(lst):
    total=0
    for i in range(len(lst)):
        total+=lst[i]
    return total
lst=[1,2,3,4,5,6,7]
print(sum_of_ele(lst))
```
